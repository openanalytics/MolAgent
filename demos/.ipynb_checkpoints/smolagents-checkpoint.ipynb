```python
import os

from smolagents import CodeAgent, OpenAIServerModel
from smolagents.mcp_client import MCPClient


# Remote MCP server
MCP_URL = os.environ["MOLAGENT_MCP_URL"]

# Keep credentials outside the source code
MCP_TOKEN = os.environ["MOLAGENT_TOKEN"]


mcp_client = MCPClient(
    {
        "url": MCP_URL,
        "headers": {
            "Authorization": f"Bearer {MCP_TOKEN}",
        },
    }
)

with mcp_client:
    tools = mcp_client.get_tools()

    model = OpenAIServerModel(
        model_id="your-model",
        api_base="https://your-llm-provider.example.com/v1",
        api_key=os.environ["LLM_API_KEY"],
    )

    agent = CodeAgent(
        tools=tools,
        model=model,
    )

    result = agent.run(
        "Use the MCP tools train a modle for data.csv"
    )

    print(result)
```
